```python
# task no : 5 

def calculate(expression):
    tokens = []
    number = ""

    for char in expression:
        if char.isdigit():
            number += char
        else:
            if number:
                tokens.append(int(number))
                number = ""

            if char in "+-*/()":
                tokens.append(char)
            elif char.isspace():
                continue
            else:
                raise ValueError("Invalid character")

    if number:
        tokens.append(int(number))

    index = 0

    def factor():
        nonlocal index

        if index >= len(tokens):
            raise ValueError("Invalid expression")

        token = tokens[index]

        if token == "-":
            index += 1
            return -factor()

        if token == "+":
            index += 1
            return factor()

        if token == "(":
            index += 1
            value = expression_value()

            if index >= len(tokens) or tokens[index] != ")":
                raise ValueError("Missing closing parenthesis")

            index += 1
            return value

        if isinstance(token, int):
            index += 1
            return token

        raise ValueError("Invalid expression")

    def term():
        nonlocal index
        value = factor()

        while index < len(tokens) and tokens[index] in ("*", "/"):
            operator = tokens[index]
            index += 1
            right = factor()

            if operator == "*":
                value *= right
            else:
                if right == 0:
                    raise ZeroDivisionError("Division by zero")
                value /= right

        return value

    def expression_value():
        nonlocal index
        value = term()

        while index < len(tokens) and tokens[index] in ("+", "-"):
            operator = tokens[index]
            index += 1
            right = term()

            if operator == "+":
                value += right
            else:
                value -= right

        return value

    result = expression_value()

    if index != len(tokens):
        raise ValueError("Malformed expression")

    return result


print(calculate("-(3 + 2) * 4 + 18 / 3"))
```

```python
# task no : 6 
from dataclasses import dataclass
from uuid import uuid4


@dataclass(frozen=True)
class Transaction:
    transaction_id: str
    action: str
    amount: float
    balance: float


class Account:
    def __init__(self, account_id, balance=0):
        self.account_id = account_id
        self._balance = balance
        self._transactions = []

    @property
    def balance(self):
        return self._balance

    @property
    def statement(self):
        return tuple(self._transactions)

    def _record(self, action, amount):
        transaction = Transaction(
            str(uuid4()),
            action,
            amount,
            self._balance
        )
        self._transactions.append(transaction)

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")

        self._balance += amount
        self._record("deposit", amount)

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")

        if amount > self._balance:
            raise ValueError("Insufficient funds")

        self._balance -= amount
        self._record("withdrawal", amount)


class Bank:
    def __init__(self):
        self.accounts = {}

    def add_account(self, account):
        self.accounts[account.account_id] = account

    def get_account(self, account_id):
        if account_id not in self.accounts:
            raise ValueError("Account does not exist")

        return self.accounts[account_id]

    def transfer(self, from_id, to_id, amount):
        source = self.get_account(from_id)
        destination = self.get_account(to_id)

        if amount <= 0:
            raise ValueError("Amount must be positive")

        if source.balance < amount:
            raise ValueError("Insufficient funds")

        source._balance -= amount
        destination._balance += amount

        source._record("transfer_out", amount)
        destination._record("transfer_in", amount)


bank = Bank()

account_a = Account("A", 1000)
account_b = Account("B", 200)

bank.add_account(account_a)
bank.add_account(account_b)

bank.transfer("A", "B", 350)

print(account_a.balance)
print(account_b.balance)

print(account_a.statement)
print(account_b.statement)
```

```python
## task no : 7 
from abc import ABC, abstractmethod


class Employee(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def calculate_pay(self):
        pass


class SalariedEmployee(Employee):
    def __init__(self, name, salary):
        super().__init__(name)
        self.salary = salary

    def calculate_pay(self):
        return self.salary


class HourlyEmployee(Employee):
    def __init__(self, name, rate, hours):
        super().__init__(name)
        self.rate = rate
        self.hours = hours

    def calculate_pay(self):
        if self.hours <= 40:
            return self.rate * self.hours

        regular_pay = 40 * self.rate
        overtime_pay = (self.hours - 40) * self.rate * 1.5

        return regular_pay + overtime_pay


class CommissionEmployee(Employee):
    def __init__(self, name, base, sales, rate):
        super().__init__(name)
        self.base = base
        self.sales = sales
        self.rate = rate

    def calculate_pay(self):
        return self.base + self.sales * self.rate


employees = [
    SalariedEmployee("Ali", 3000),
    HourlyEmployee("Sara", 20, 45),
    CommissionEmployee("Qasim", 500, 4000, 0.10)
]

payroll_summary = []

for employee in employees:
    payroll_summary.append({
        "name": employee.name,
        "pay": employee.calculate_pay()
    })

print(payroll_summary)
```

```python
## task no : 8 
import time
import functools


def ttl_cache(seconds):
    def decorator(func):
        cache = {}
        hits = 0
        misses = 0

        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            nonlocal hits, misses

            key = (args, tuple(sorted(kwargs.items())))

            try:
                hash(key)
            except TypeError:
                return func(*args, **kwargs)

            now = time.monotonic()

            if key in cache:
                result, expiry = cache[key]

                if now < expiry:
                    hits += 1
                    return result

            misses += 1
            result = func(*args, **kwargs)

            cache[key] = (result, now + seconds)

            return result

        wrapper.cache_hits = lambda: hits
        wrapper.cache_misses = lambda: misses

        return wrapper

    return decorator


@ttl_cache(seconds=5)
def add(x, y):
    print("Calculating...")
    return x + y


print(add(x=2, y=3))
print(add(y=3, x=2))

print("Hits:", add.cache_hits())
print("Misses:", add.cache_misses())
```

```python
## task no : 9
from collections import defaultdict
from datetime import datetime


def read_errors(path):
    read_errors.malformed_count = 0

    with open(path, "r", encoding="utf-8") as file:
        for line_number, line in enumerate(file, start=1):
            parts = line.strip().split(" | ", 2)

            if len(parts) != 3:
                read_errors.malformed_count += 1
                continue

            timestamp, level, message = parts

            try:
                datetime.strptime(timestamp, "%Y-%m-%d %H:%M:%S")
            except ValueError:
                read_errors.malformed_count += 1
                continue

            if not level or not message:
                read_errors.malformed_count += 1
                continue

            if level == "ERROR":
                yield line_number, timestamp, message


def summarize_errors(path):
    summary = defaultdict(int)

    for line_number, timestamp, message in read_errors(path):
        date = timestamp.split()[0]
        summary[date] += 1

    return dict(summary)


# Create sample log file
with open("app.log", "w", encoding="utf-8") as file:
    file.write("2026-11-07 10:30:00 | ERROR | Connection lost\n")
    file.write("2026-11-07 11:00:00 | INFO | Application started\n")
    file.write("2026-11-07 12:00:00 | ERROR | Database failed\n")
    file.write("Invalid log line\n")


# Read errors
for error in read_errors("app.log"):
    print(error)

print("Malformed lines:", read_errors.malformed_count)

# Summarize errors
print(summarize_errors("app.log"))
```

```python
## task no : 10 

import csv

# Create inventory.csv
with open("inventory.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["sku", "name", "quantity"])
    writer.writerow(["P01", "Laptop", 10])
    writer.writerow(["P02", "Mouse", 20])

# Create movements.csv
with open("movements.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["sku", "type", "units"])
    writer.writerow(["P01", "OUT", 4])
    writer.writerow(["P01", "OUT", 9])
    writer.writerow(["P01", "IN", 2])
    writer.writerow(["P02", "IN", 5])
    writer.writerow(["P99", "IN", 3])
    writer.writerow(["P02", "OUT", -2])


# Read inventory
inventory = {}

with open("inventory.csv", "r", newline="") as file:
    reader = csv.DictReader(file)

    for row in reader:
        inventory[row["sku"]] = {
            "name": row["name"],
            "quantity": int(row["quantity"])
        }


# Process movements
rejected = []

with open("movements.csv", "r", newline="") as file:
    reader = csv.DictReader(file)

    for row in reader:
        sku = row["sku"]
        movement = row["type"]

        try:
            units = int(row["units"])
        except ValueError:
            rejected.append([sku, movement, row["units"], "Invalid units"])
            continue

        if sku not in inventory:
            rejected.append([sku, movement, units, "Unknown SKU"])
            continue

        if units <= 0:
            rejected.append([sku, movement, units, "Units must be positive"])
            continue

        if movement not in ["IN", "OUT"]:
            rejected.append([sku, movement, units, "Invalid movement type"])
            continue

        if movement == "OUT":
            if inventory[sku]["quantity"] < units:
                rejected.append([sku, movement, units, "Insufficient stock"])
                continue

            inventory[sku]["quantity"] -= units

        elif movement == "IN":
            inventory[sku]["quantity"] += units


# Write updated inventory
with open("updated_inventory.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["sku", "name", "quantity"])

    for sku, data in inventory.items():
        writer.writerow([sku, data["name"], data["quantity"]])


# Write rejected movements
with open("rejected_movements.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["sku", "type", "units", "reason"])
    writer.writerows(rejected)


print("Inventory reconciliation complete!")
print("Updated inventory:", inventory)
print("Rejected movements:", rejected)
```

```python
## task no : 11 
import csv

# Create students.csv
with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["student_id", "name"])
    writer.writerow([101, "Ali"])
    writer.writerow([102, "Sara"])
    writer.writerow([103, "Ahmed"])


# Create scores.csv
with open("scores.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["student_id", "subject", "marks"])
    writer.writerow([101, "Math", 90])
    writer.writerow([101, "English", 80])
    writer.writerow([101, "Science", 85])
    writer.writerow([102, "Math", 75])
    writer.writerow([102, "English", 65])
    writer.writerow([103, "Math", 45])


# Read students
students = {}

with open("students.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        students[row["student_id"]] = {
            "name": row["name"],
            "marks": []
        }


# Read scores
seen = set()
rejected = []

with open("scores.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        student_id = row["student_id"]
        subject = row["subject"]

        try:
            marks = float(row["marks"])
        except ValueError:
            rejected.append(row)
            continue

        if student_id not in students:
            rejected.append(row)
            continue

        if (student_id, subject) in seen:
            rejected.append(row)
            continue

        if marks < 0 or marks > 100:
            rejected.append(row)
            continue

        seen.add((student_id, subject))
        students[student_id]["marks"].append(marks)


# Calculate average and grade
report = []

for student_id, data in students.items():

    if data["marks"]:
        average = sum(data["marks"]) / len(data["marks"])
    else:
        average = 0

    if average >= 85:
        grade = "A"
    elif average >= 70:
        grade = "B"
    elif average >= 55:
        grade = "C"
    elif average >= 40:
        grade = "D"
    else:
        grade = "F"

    report.append({
        "student_id": student_id,
        "name": data["name"],
        "average": round(average, 2),
        "grade": grade
    })


# Sort by average descending, then ID ascending
report.sort(key=lambda x: (-x["average"], x["student_id"]))


# Assign ranks
for rank, student in enumerate(report, start=1):
    student["rank"] = rank


# Export report.csv
with open("report.csv", "w", newline="") as file:
    writer = csv.DictWriter(
        file,
        fieldnames=["student_id", "name", "average", "grade", "rank"]
    )

    writer.writeheader()
    writer.writerows(report)


print("Gradebook complete!")
print(report)
print("Rejected records:", rejected)
```

```python
## rask no : 12 
class ReversibleRange:

    def __init__(self, start, stop, step=1):
        if step == 0:
            raise ValueError("Step cannot be zero")

        self.start = start
        self.stop = stop
        self.step = step

    def __iter__(self):
        current = self.start

        if self.step > 0:
            while current < self.stop:
                yield current
                current += self.step

        else:
            while current > self.stop:
                yield current
                current += self.step

    def __reversed__(self):
        return reversed(range(self.start, self.stop, self.step))

    def __len__(self):
        return len(range(self.start, self.stop, self.step))

    def __contains__(self, value):
        return value in range(self.start, self.stop, self.step)



numbers = ReversibleRange(2, 11, 3)

print(list(numbers))
print(list(reversed(numbers)))
print(len(numbers))
print(5 in numbers)
print(6 in numbers)
```

```python
## task no : 13
from datetime import datetime


class BookingScheduler:
    def __init__(self):
        self.bookings = []

    def add_booking(self, room, start, end, booking_id):

        if end <= start:
            return "Rejected: Invalid booking time"

        for booking in self.bookings:
            if booking["booking_id"] == booking_id:
                return "Rejected: Duplicate booking ID"

            if booking["room"] == room:
                if start < booking["end"] and end > booking["start"]:
                    return "Rejected: Booking conflict"

        self.bookings.append({
            "room": room,
            "start": start,
            "end": end,
            "booking_id": booking_id
        })

        self.bookings.sort(key=lambda x: (x["room"], x["start"]))

        return "Booking accepted"


scheduler = BookingScheduler()

print(scheduler.add_booking(
    "A",
    datetime(2026, 11, 7, 9, 0),
    datetime(2026, 11, 7, 10, 0),
    1
))

print(scheduler.add_booking(
    "A",
    datetime(2026, 11, 7, 10, 0),
    datetime(2026, 11, 7, 11, 0),
    2
))

print(scheduler.add_booking(
    "A",
    datetime(2026, 11, 7, 9, 30),
    datetime(2026, 11, 7, 10, 30),
    3
))
```

```python
## task no : 14
import json
import os

FILE_NAME = "contacts.json"


def load_contacts():
    if not os.path.exists(FILE_NAME):
        return []

    try:
        with open(FILE_NAME, "r") as file:
            data = json.load(file)

        if not isinstance(data, list):
            raise ValueError("Invalid JSON data")

        return data

    except (json.JSONDecodeError, ValueError):
        print("Contacts file is corrupted!")
        raise SystemExit


def save_contacts(contacts):
    with open(FILE_NAME, "w") as file:
        json.dump(contacts, file, indent=4)


contacts = load_contacts()


def add_contact():
    contact_id = input("Enter ID: ")

    for contact in contacts:
        if contact["id"] == contact_id:
            print("ID already exists!")
            return

    name = input("Enter name: ")
    phone = input("Enter phone: ")
    email = input("Enter email: ")

    if not name.strip() or not phone.strip() or "@" not in email:
        print("Invalid contact details!")
        return

    contacts.append({
        "id": contact_id,
        "name": name,
        "phone": phone,
        "email": email
    })

    save_contacts(contacts)
    print("Contact added successfully!")


def search_contact():
    name = input("Enter name to search: ").lower()
    found = False

    for contact in contacts:
        if name in contact["name"].lower():
            print(contact)
            found = True

    if not found:
        print("Contact not found!")


def update_contact():
    contact_id = input("Enter contact ID: ")

    for contact in contacts:
        if contact["id"] == contact_id:
            name = input("Enter new name: ")
            phone = input("Enter new phone: ")
            email = input("Enter new email: ")

            if not name.strip() or not phone.strip() or "@" not in email:
                print("Invalid details!")
                return

            contact["name"] = name
            contact["phone"] = phone
            contact["email"] = email

            save_contacts(contacts)
            print("Contact updated!")
            return

    print("Contact not found!")


def delete_contact():
    contact_id = input("Enter contact ID: ")

    for contact in contacts:
        if contact["id"] == contact_id:
            contacts.remove(contact)
            save_contacts(contacts)
            print("Contact deleted!")
            return

    print("Contact not found!")


def list_contacts():
    if not contacts:
        print("No contacts available!")
        return

    for contact in contacts:
        print(contact)


def menu():
    while True:
        print("\n===== CONTACT MANAGER =====")
        print("1. Add Contact")
        print("2. Search Contact")
        print("3. Update Contact")
        print("4. Delete Contact")
        print("5. List Contacts")
        print("6. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_contact()
        elif choice == "2":
            search_contact()
        elif choice == "3":
            update_contact()
        elif choice == "4":
            delete_contact()
        elif choice == "5":
            list_contacts()
        elif choice == "6":
            break
        else:
            print("Invalid choice!")


menu()
```

```python
## task no : 15 
import json
import os
import uuid
import unittest


CATALOG_FILE = "products.json"
ORDERS_FILE = "orders.json"


class Product:
    def __init__(self, product_id, name, price, stock):
        if price < 0 or stock < 0:
            raise ValueError("Price and stock cannot be negative")

        self.product_id = product_id
        self.name = name
        self.price = price
        self.stock = stock


class Cart:
    def __init__(self):
        self.items = {}

    def add_product(self, product, quantity):
        if quantity <= 0:
            raise ValueError("Quantity must be positive")

        self.items[product.product_id] = (
            self.items.get(product.product_id, 0) + quantity
        )

    def calculate_total(self, products, discount=0):
        if not 0 <= discount <= 100:
            raise ValueError("Invalid discount")

        total = 0

        for product_id, quantity in self.items.items():
            product = products[product_id]
            total += product.price * quantity

        return total - (total * discount / 100)


class Inventory:
    def __init__(self, products):
        self.products = products

    def validate_items(self, cart):
        for product_id, quantity in cart.items.items():
            if quantity <= 0:
                raise ValueError("Invalid quantity")

            if product_id not in self.products:
                raise ValueError("Product not found")

            if self.products[product_id].stock < quantity:
                raise ValueError("Out of stock")

    def update_stock(self, cart):
        for product_id, quantity in cart.items.items():
            self.products[product_id].stock -= quantity


class Order:
    def __init__(self, cart, products, discount):
        self.order_id = str(uuid.uuid4())
        self.items = cart.items.copy()
        self.total = cart.calculate_total(products, discount)

    def to_dict(self):
        return {
            "order_id": self.order_id,
            "items": self.items,
            "total": self.total
        }


class OrderFulfillmentSystem:
    def __init__(self):
        self.products = self.load_catalog()
        self.inventory = Inventory(self.products)

    def load_catalog(self):
        if not os.path.exists(CATALOG_FILE):
            sample = [
                {
                    "product_id": "P1",
                    "name": "Product 1",
                    "price": 100,
                    "stock": 5
                }
            ]

            with open(CATALOG_FILE, "w") as file:
                json.dump(sample, file, indent=4)

        with open(CATALOG_FILE, "r") as file:
            data = json.load(file)

        products = {}

        for item in data:
            product = Product(
                item["product_id"],
                item["name"],
                item["price"],
                item["stock"]
            )
            products[product.product_id] = product

        return products

    def save_order(self, order):
        if os.path.exists(ORDERS_FILE):
            with open(ORDERS_FILE, "r") as file:
                orders = json.load(file)
        else:
            orders = []

        orders.append(order.to_dict())

        with open(ORDERS_FILE, "w") as file:
            json.dump(orders, file, indent=4)

    def place_order(self, cart, discount=0):
        self.inventory.validate_items(cart)

        order = Order(cart, self.products, discount)

        old_stock = {
            product_id: product.stock
            for product_id, product in self.products.items()
        }

        try:
            self.inventory.update_stock(cart)
            self.save_order(order)

        except Exception:
            for product_id, stock in old_stock.items():
                self.products[product_id].stock = stock

            raise

        return order


# ---------------- UNIT TESTS ----------------

class TestOrderSystem(unittest.TestCase):

    def setUp(self):
        self.products = {
            "P1": Product("P1", "Product 1", 100, 5)
        }
        self.inventory = Inventory(self.products)

    def test_valid_order(self):
        cart = Cart()
        cart.add_product(self.products["P1"], 2)

        self.inventory.validate_items(cart)

        self.assertEqual(
            cart.calculate_total(self.products, 10),
            180
        )

    def test_invalid_quantity(self):
        cart = Cart()

        with self.assertRaises(ValueError):
            cart.add_product(self.products["P1"], 0)

    def test_out_of_stock(self):
        cart = Cart()
        cart.add_product(self.products["P1"], 10)

        with self.assertRaises(ValueError):
            self.inventory.validate_items(cart)


# ---------------- RUN PROGRAM ----------------

if __name__ == "__main__":

    system = OrderFulfillmentSystem()

    cart = Cart()
    cart.add_product(system.products["P1"], 2)

    try:
        order = system.place_order(cart, discount=10)

        print("Order placed successfully!")
        print("Order ID:", order.order_id)
        print("Total:", order.total)
        print("Remaining stock:", system.products["P1"].stock)

    except Exception as error:
        print("Order failed:", error)

    print("\nRunning Unit Tests...")
    unittest.main(argv=["first-arg-is-ignored"], exit=False)
```
